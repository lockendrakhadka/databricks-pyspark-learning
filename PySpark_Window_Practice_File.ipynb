```python
# Creating DataFrame for employee_details 
emp_df = spark.read.table("workspace.company_db.employee_details")
```

```python
# Import functions 
from pyspark.sql.window import Window
from pyspark.sql.functions import col, row_number
```

```python
# Window specification and order salary by descending
window_spec = Window.partitionBy("department_id").orderBy(
    col("salary").desc()
)
```

```python
# Window Function
emp_ranked = emp_df.withColumn(
    "row_number",
    row_number().over(window_spec)
)
```

```python
emp_ranked.show()
```

WE are continuing with row_number since we have different values in salary 

```python
# Filter Top Employee 
top_employee = emp_ranked.filter(col("row_number") == 1)
```

```python
# Check the Top salary Employee
top_employee.show()
```

```python
# Using sum() as a window function
from pyspark.sql.functions import sum

total_window = Window.partitionBy("department_id")

emp_salary = emp_df.withColumn(
    "department_total_salary",
    sum("salary").over(total_window)
)
```

```python
emp_salary.show()
```

```python
# Avg() salary per department 
from pyspark.sql.functions import avg

emp_avg = emp_df.withColumn(
    "department_avg_salary",
    avg("salary").over(total_window)
)
```

```python
emp_avg.show()
```

```python
# Max() salary per department
from pyspark.sql.functions import max

emp_max = emp_df.withColumn(
    "department_max_salary",
    max("salary").over(total_window)
)
```

```python
emp_max.show()
```

```python
# Min() salary per department
from pyspark.sql.functions import min

emp_min = emp_df.withColumn(
    "department_min_salary",
    min("salary").over(total_window)
)
```

```python
emp_min.show()
```

```python
# Count emp per department
from pyspark.sql.functions import count

department_count = emp_df.groupBy("department_id").count()
```

```python
department_count.show()
```
