```python
# Creating a table in the company_db database 
spark.sql("""
CREATE TABLE workspace.company_db.employee_details (
    employee_id INT,
    employee_name STRING,
    department_id INT,
    job_title STRING,
    salary DOUBLE,
    age INT,
    joining_date DATE
)
""")
```

```python
# Creating another database 
spark.sql("""
CREATE SCHEMA IF NOT EXISTS workspace.hr_db
""")
```

```python
# Creating table
spark.sql("""
CREATE TABLE workspace.hr_db.departments (
    department_id INT,
    department_name STRING,
    location STRING
)
""")
```

```python
# Insert values
spark.sql("""
INSERT INTO workspace.hr_db.departments
VALUES
(1, 'IT', 'New Delhi'),
(2, 'HR', 'Mumbai'),
(3, 'Finance', 'Bangalore'),
(4, 'Sales', 'Pune'),
(5, 'Marketing', 'Hyderabad')
""")
```

```python
# Join employee_details from company_db with departments from hr_db
joined_tables = spark.sql("""
SELECT
    e.employee_id,
    e.employee_name,
    e.department_id,
    d.department_name,
    d.location,
    e.job_title,
    e.salary
FROM workspace.company_db.employee_details e
JOIN workspace.hr_db.departments d
    ON e.department_id = d.department_id
""")
```
