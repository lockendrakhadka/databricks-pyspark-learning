```python

from pyspark.sql import functions as F

# ADLS storage account
STORAGE_ROOT = (
    "abfss://lakehouse@storagelokiiiiiiii"
    ".dfs.core.windows.net"
)

# Source Delta tables and their ADLS output folders
TABLES = [
    "customers",
    "products",
    "orders",
    "employees",
    "departments"
]

for table_name in TABLES:
    print(f"Exporting {table_name}...")

    # Read the latest processed data
    target_table = f"target_catalog.curated.{table_name}"
    df = spark.table(target_table)

    # Output folder in ADLS
    output_path = (
        f"{STORAGE_ROOT}/processed/{table_name}"
    )

    # Write the current table data as CSV
    (
        df.coalesce(1)
        .write
        .format("csv")
        .option("header", "true")
        .mode("overwrite")
        .save(output_path)
    )

    print(f"Successfully exported {table_name}")
    print(f"Output: {output_path}")

print("All five tables exported successfully.")
```
