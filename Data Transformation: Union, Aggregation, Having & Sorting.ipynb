```python
# Reading Csv file 
df_branch1 = spark.read.csv("/Volumes/workspace/company_db/emp/sales_branch1.csv",inferSchema=True,
    header=True)

df_branch2 = spark.read.csv("/Volumes/workspace/company_db/emp/sales_branch2.csv",inferSchema=True,
    header=True)
```

```python
df_branch1.show()
df_branch2.show()
```

```python
# Step 2 — Union the two DataFrames

df_sales = df_branch1.union(df_branch2)

df_sales.show()
```

```python
# Step 3 — Aggregation

from pyspark.sql.functions import sum

total_sales_category = df_sales.groupBy("category").agg(sum("amount").alias("total_sales"))

total_sales_category.show()
```

```python
# Step 4 — Having after group by

filtered_sales = total_sales_category.filter(total_sales_category.total_sales > 1500)

filtered_sales.show()
```

```python
# Step 5 — Sorting total sales asc

sorted_sales = total_sales_category.orderBy("total_sales")

sorted_sales.show()
```

```python
# Step 6 — Sorting total sales desc

desc_sales = total_sales_category.orderBy("total_sales", ascending=False)

desc_sales.show()
```
