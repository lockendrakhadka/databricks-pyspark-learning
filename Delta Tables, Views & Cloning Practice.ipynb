**Creating a delta table using if table is not exist**

```python
%sql

CREATE TABLE IF NOT EXISTS workspace.company_db.employee_practice_delta (
    emp_id INT,
    emp_name STRING,
    dept_id INT,
    salary DOUBLE
)
USING DELTA;
```

**Now we have created a table and if we try to create the table with same name again without using IF NOT EXISTS it will show an error**
