```python
# Reading the data

df = spark.read.csv("/Workspace/Users/lockendrakhadka222@gmail.com/databricks-pyspark-learning/Assignment Folder/EmployeeData.csv", header=True, inferSchema=True)
```

<b> How many employees have a first name that starts with the letter 'J'?

```python
# Write your code using the Spark Function

from pyspark.sql.functions import col, lower


j_employees = df.filter(lower(col("first_name")).like("j%")).count()


print("Employees whose first name starts with J:", j_employees)
```

<b> How many employees have a last name that ends with the letter 's'?

```python
# Write your code using the Spark Function

s_employees = df.filter(lower(col("last_name")).like("%s")).count()


print("Employees whose Last name ends with s:", s_employees)
```

<b> What is the length of the longest first name?

```python
# Write your code using the Spark Function


from pyspark.sql.functions import length,max

longest_first_name = df.select(max(length("first_name")).alias("Longest_first_name_length"))

longest_first_name.show()
```

d
<b> What is the length of the shortest last name?

```python
# Write your code using the Spark Function

from pyspark.sql.functions import min

shortest_last_name = df.select(min(length("last_name")).alias("shortest_last_name_length"))

shortest_last_name.show()
```

<b> What is the email domain of the employee with the highest salary?

```python
# Write your code using the Spark Function


from pyspark.sql.functions import split


highest_salary = df.orderBy(df.salary.desc()).limit(1)


result = highest_salary.select(split(col("email"),"@").getItem(1).alias("Email_domain"))


result.show()
```

<b> What is the first name of the employee with the email address that contains the substring 'hr'?

```python
# Write your code using the Spark Function


email_hr = df.filter(col("email").like("%hr%")& col("first_name").isNotNull()).select("first_name")

email_hr.show()
```

<b> What is the last name of the employee with the email address that ends with the domain '.com'?

```python
# Write your code using the Spark Function


email_end = df.filter(col("email").like("%.com")).select("last_name")

email_end.show()
```

<b> How many employees have an email address that contains their last name?

```python
# Write your code using the Spark Function


lastname_email = df.filter(lower(col("email")).contains(lower(col("last_name")))).count()


print("Employees whose email contains their last name:", lastname_email)
```

<b> What is the email of the employee with the first name that contains a space character?

```python
# Write your code using the Spark Function


space_in_name = df.filter(col("first_name").like("% %")).select("first_name","email")

space_in_name.show()
```

```python
# Write your code using the Spark Function

hyphen_in_name = df.filter(col("last_name").like("%-%")).select("last_name","email")

hyphen_in_name.show()
```
