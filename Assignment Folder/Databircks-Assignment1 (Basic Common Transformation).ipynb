<b>How many employees are in the input file?

```python
# Write your code using the Spark Function
df = spark.read.csv("/Workspace/Users/lockendrakhadka222@gmail.com/databricks-pyspark-learning/Assignment Folder/EmployeeData.csv", header="true", inferSchema="true")

count=df.count()

print("Total number of employee:", count)
```

<b> What is the average salary of all the employees?

```python
# Write your code using the Spark Function

from pyspark.sql.functions import avg

average_salary=df.select(avg("salary").alias("avg_salary")) 

average_salary.show()
```

<b> How many male and female employees are there in the input file?

```python
# Write your code using the Spark Function

gender_count=df.groupBy("gender").count()
gender_count.show()
```

<b> Who is the employee with the highest salary?

```python
# Write your code using the Spark Function

max_salary=df.orderBy(df.salary.desc()).limit(1)
max_salary.select("first_name", "last_name", "salary").show()
```

<b> What is the total bonus amount paid to all the employees?

```python
# Write your code using the Spark Function

from pyspark.sql.functions import sum, format_number

bonus_total=df.select(format_number(sum("bonus"),2).alias("Bonus_total"))
bonus_total.show()
```

<b> How many employees were hired in the year 2020?

```python
# Write your code using the Spark Function

from pyspark.sql.functions import year

year_2020=df.filter(year("creationDate")==2020).count()

print("Employees hired in 2020:", year_2020)
```
