```python
# reading/loading a CSV file into a DataFrame
df = spark.read.csv(
    "/Volumes/workspace/company_db/emp",
    header=True,
    inferSchema=False
)
```

```python
# to view
df.show()
```

```python
# Convert data type 
from pyspark.sql.functions import col, to_date

df_correct=df.select(
    col("employee_id").cast("int"),
    col("employee_name"),
    col("department"),
    col("job_title"),
    col("salary").cast("double"),
    col("age").cast("int"),
    to_date(col("joining_date"),"yyyy-MM-dd").alias("joining_date")
)
```

```python
#Check schema
df_correct.printSchema()
```

```python
# view data
df_correct.show()
```

Transformations are done below

```python
# Transformation 1 — Filter employees with the certaian condition
df_filtered = df_correct.filter(df_correct.salary > 50000)
```

```python
# View data
df_filtered.show()
```

```python
# Transformation 2 — Select specific column
df_select=df_correct.select("employee_id","employee_name","department","salary")
```

```python
# View data
df_select.show()
```

```python
# Transformation 3 — Create new column
df_annual = df_correct.withColumn(
    "annual_salary",
    col("salary") * 12
)
```

```python
# View data
df_annual.show()
```

```python
# Transformation 4 — Remove specific column
df_without_age = df_correct.drop("age")
```

```python
# View data
df_without_age.show()
```

```python
# Transformation — Useing groupBy()
df_grouped = df_correct.groupBy("department").count()
```

```python
# View data
df_grouped.show()
```
