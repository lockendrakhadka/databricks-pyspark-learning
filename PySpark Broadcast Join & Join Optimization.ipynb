**We will be using 3 tables here 2 Big tables to perform sort merge join and 1 small table to perform broadcast join, department table here is a small table**

```python
# Step 1 — Read files

df_employees = spark.read.csv("/Volumes/workspace/company_db/emp/employee_large.csv",
    header=True,
    inferSchema=True)

df_departments = spark.read.csv("/Volumes/workspace/company_db/emp/department (1).csv",
    header=True,
    inferSchema=True)


employee_details= spark.read.csv("/Volumes/workspace/company_db/emp/employee_detail_large.csv",
    header=True,
    inferSchema=True)
```

```python
# Check the dataframe

df_employees.show()
df_departments.show()
employee_details.show()
```

**We are telling the spark to choose sort merge join here because it automatically chooses broadcast join**

```python
# Step 2 — Perform a sort merge join


df_sort_merge = df_employees.hint("merge").join(
    employee_details.hint("merge"),
    df_employees.emp_id == employee_details.emp_id,
    "inner"
)
```

```python
# We are checking what execution plan will spark choose

df_sort_merge.explain(True)
```

```python
# check the count and see the performance and how the spark actually process the sort merge join 

df_sort_merge.count()
```

**We will perform broadcast join below**

```python
# Since above we have given it a hind to perform sort merge join we will now say it to use broad cast join

from pyspark.sql.functions import broadcast

df_broadcast = df_employees.join(
    broadcast(df_departments),
    df_employees.dept_id == df_departments.dept_id,
    "inner"
)
```

```python
# We are checking what execution plan will spark choose


df_broadcast.explain(True)
```

```python
# check the count and see the performance and how the spark actually process the Broadcast join 

df_broadcast.count()
```
