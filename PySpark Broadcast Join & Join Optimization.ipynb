```python
# Step 1 — Read files

df_employees = spark.read.csv("/Volumes/workspace/company_db/emp/employees_broadcast_join.csv",
    header=True,
    inferSchema=True)

df_departments = spark.read.csv("/Volumes/workspace/company_db/emp/departments_broadcast_join.csv",
    header=True,
    inferSchema=True)
```

```python
# Check the dataframe

df_employees.show()
df_departments.show()
```

```python
# Step 2 — Perform a normal join

df_normal_join = df_employees.join(df_departments,df_employees.department_id ==df_departments.department_id,"inner")
```
