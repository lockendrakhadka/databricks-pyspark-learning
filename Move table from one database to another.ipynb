```python
# Copy table from hr_db to company_db as dept_df DataFrame
dept_df = spark.read.table("workspace.hr_db.departments")

dept_df.write.saveAsTable("workspace.company_db.departments")
```

```python
# First, read both tables as DataFrames:
emp_df = spark.read.table("workspace.company_db.employee_details")
dept_df = spark.read.table("workspace.company_db.departments")
```

```python
# Join the tables and select the column
joined_df = emp_df.join(
    dept_df,
    emp_df.department_id == dept_df.department_id,
    "inner"
).select(
    emp_df.employee_id,
    emp_df.employee_name,
    emp_df.job_title,
    emp_df.salary,
    emp_df.age,
    emp_df.joining_date,
    emp_df.department_id,
    dept_df.department_name,
    dept_df.location
)
```
